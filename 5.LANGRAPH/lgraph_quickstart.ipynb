```python
from langgraph.graph import StateGraph,END
from typing import TypedDict#TypedDict is commonly used in LangGraph to define the structure of the State.
from IPython.display import Image,display
```

```python
class Mystate(TypedDict): #TypedDict allows you to define the expected keys and their types.
    message:str

#Define the start node
def start_node(state:Mystate)-> Mystate:
    #state: Mystate
    #This means the function expects a state object whose structure follows Mystate.
    print("Start the node received:",state['message'])
    #process or pass the message to the end node
    return {"message":state["message"]+" --> processed by start"}

#define the end node
def end_node(state:Mystate)-> Mystate:
    print("End node received:",state["message"])
    return state


#Build the graph
builder=StateGraph(Mystate)

```

```python
id(builder)
```

```python
builder
```

```python
#Add nodes
builder.add_node("start",start_node)
builder.add_node("end",end_node)
```

```python
#Define the flow
builder.set_entry_point("start")
builder.add_edge("start","end")
builder.add_edge("end",END) #END is a special marker for graph completion
```

```python
#compile the graph
graph=builder.compile()

#run the graph
result=graph.invoke({"message":"what is lanchain vs langgraph"})
print("FInal result",result["message"])

#invoke() means:Start executing the LangGraph with this initial State.
```
