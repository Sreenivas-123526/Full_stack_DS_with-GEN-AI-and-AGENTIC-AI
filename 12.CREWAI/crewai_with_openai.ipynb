```python
import os

from crewai import Agent, Task, Crew, Process, LLM
from crewai_tools import SerperDevTool


# --------------------------------------------------
# API keys
# --------------------------------------------------

os.environ["OPENAI_API_KEY"] = "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"
os.environ["SERPER_API_KEY"] = "8e079e06b931547c0647a15e3e3ecb17b3d899e1"


# --------------------------------------------------
# Search tool
# --------------------------------------------------

search_tool = SerperDevTool()


# --------------------------------------------------
# LLM
# --------------------------------------------------

llm = LLM(
    model="openai/gpt-5.4",
    temperature=0.7
)


# --------------------------------------------------
# Researcher Agent
# --------------------------------------------------

researcher = Agent(
    role="Senior Research Analyst",

    goal="Uncover cutting-edge developments in AI and data science",

    backstory="""
    You work at a leading tech think tank.
    Your expertise lies in identifying emerging trends.
    You have a knack for dissecting complex data
    and presenting actionable insights.
    """,

    verbose=True,

    allow_delegation=False,

    llm=llm,

    tools=[search_tool]
)


# --------------------------------------------------
# Writer Agent
# --------------------------------------------------

writer = Agent(
    role="Tech Content Strategist",

    goal="Craft compelling content on tech advancements",

    backstory="""
    You are a renowned Content Strategist, known for
    your insightful and engaging articles.

    You transform complex concepts into compelling narratives.
    """,

    verbose=True,

    allow_delegation=True,

    llm=llm
)


# --------------------------------------------------
# Task 1
# --------------------------------------------------

task1 = Task(
    description="""
    Conduct a comprehensive analysis of the latest
    advancements in AI.

    Identify:
    - Key trends
    - Breakthrough technologies
    - Potential industry impacts
    """,

    expected_output="""
    Full analysis report in bullet points.
    """,

    agent=researcher
)


# --------------------------------------------------
# Task 2
# --------------------------------------------------

task2 = Task(
    description="""
    Using the insights provided by the researcher,
    develop an engaging blog post that highlights
    the most significant AI advancements.

    The post should:
    - Be informative
    - Be easy to understand
    - Be suitable for a tech-savvy audience
    - Avoid unnecessarily complex words
    """,

    expected_output="""
    Full blog post of at least 4 paragraphs.
    """,

    agent=writer
)


# --------------------------------------------------
# Crew
# --------------------------------------------------

crew = Crew(
    agents=[researcher, writer],

    tasks=[task1, task2],

    verbose=True,

    process=Process.sequential
)


# --------------------------------------------------
# Run
# --------------------------------------------------

result = await crew.kickoff_async()

print("######################")
print(result)
```
