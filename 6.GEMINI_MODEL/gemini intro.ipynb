```python
import pathlib
import textwrap
import google.generativeai as genai
from IPython.display import display
from IPython.display import Markdown



def to_markdown(text):  #The purpose is to take normal text and prepare it for Markdown display.
    text = text.replace('.', ' *')
    return Markdown(textwrap.indent(text, '> ', predicate=lambda _: True))
```

```python
for m in genai.list_models():
    if 'generateContent' in m.supported_generation_methods:
        print(m.name)
```

```python
%%time
response=model.generate_content("what is MCP in agentic AI how it works in 5 points")
```
