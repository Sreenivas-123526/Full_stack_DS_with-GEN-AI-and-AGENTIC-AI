```python
class student:
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2
        
    def __add__(self, other): ## WE DEFINE THE ADD METHOD . THIS CONCEPT IS CALLED OPERATOR OVERLOADING
        m1 = self.m1 + other.m1
        m2 = self.m2 + other.m2
        s3 = student(m1,m2)
        
        return s3   
 
s1 = student(60,70)
s2 = student(80,90)

s3 = s1 + s2 

print(s3.m1)
print(s3.m2)
```

```python
class demo():
    
    def add(self,a,b):
        return a+b
        
    def add(self, a, b, c):
        return a + b + c

    def add(self, a, b, c, d):
        return a + b + c + d

d = demo()
print(d.add(2,3))
print(d.add(2,3,4))
print(d.add(2,3, 4, 5))
```

```python
class demo():
    
    def add(self,a, b, c=0, d=0):
        return a+b+c+d 

d = demo()
print(d.add(2,3))
print(d.add(2,3,4))
print(d.add(2,3, 4, 5))
```

```python
class demo():
    
    def add(self, *arg):
        total = 0
        for i in arg:
            total = total + i
        return total

d = demo()

print(d.add(2,3))
print(d.add(4,5,6))
print(d.add(1,2,3,4,5,6,))
```

```python
class student:
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2
        
    def sum(self, a=None, b=None,c=None):
        if a!=None and b!=None and c!=None:
            s = a + b +c
        elif a!=None and b!=None:
            s = a+b
        elif a!= None and c!= None:
            s = a + c
        else:
            s = a
            
        return s
    
s1 = student(50, 60)

print(s1.sum(10)) # hear we can pass 1 parameter
print(s1.sum(10,20)) # hear we can pass 2 parameter
print(s1.sum(10,20,30)) # hear we can pass 3 parameter
print(s1.sum(10,30))

#print(s1.sum(10,20,30,40))

# THIS CONCEPT CALLED AS METHOD OVERLOADING
```
