```python
# I WANT TO PRINT TOP 10 VALUE usint 2 method iter() & next()

class topten():
    
    def __init__(self):
        self.num = 1
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.num <= 10:
            val = self.num
            self.num += 1
            return val
        else:
            raise StopIteration      # only way to stop the for loop is raise the exception          
    
values = topten()

#print(next(values))

for i in values:
    print(i)  
```

```python
def topten():
    n = 1
    
    while n<=10:
        sq = n*n
        yield sq
        n += 1
    
values = topten()
    
for i in values:
    print(i)
```
